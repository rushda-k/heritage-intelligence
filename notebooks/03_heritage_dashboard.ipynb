```python
# ============================================================
# HERITAGE INTELLIGENCE
# FINAL DASHBOARD DATA PREPARATION
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# LOAD FINAL SCORECARD
# ------------------------------------------------------------

scorecard = pd.read_csv(
    "../data/processed/heritage_intelligence_scorecard.csv"
)

gaps = pd.read_csv(
    "../data/processed/heritage_comparative_gaps.csv"
)

print("=" * 80)
print("HERITAGE INTELLIGENCE — FINAL DASHBOARD")
print("=" * 80)

print("\nMuseums analyzed:")
for museum in scorecard["museum"]:
    print("→", museum)

print("\nRecords analyzed:")
print(
    f"{scorecard['objects'].sum():,}"
)

print("\nFinal scorecard loaded successfully.")
```

```python
# ============================================================
# HERITAGE INTELLIGENCE — FINAL DASHBOARD
# ============================================================

# ------------------------------------------------------------
# 1. OVERALL SCORE
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

plt.bar(
    scorecard["museum"],
    scorecard["heritage_intelligence_score"]
)

plt.title("Heritage Intelligence Score")
plt.ylabel("Score / 100")
plt.ylim(0, 100)
plt.xticks(rotation=15)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 2. QUALITY DIMENSIONS
# ------------------------------------------------------------

quality_cols = [
    "metadata_quality_score",
    "digital_access_score",
    "artist_coverage_score",
    "culture_coverage_score",
    "country_coverage_score"
]

quality_labels = [
    "Metadata",
    "Digital Access",
    "Artist",
    "Culture",
    "Country"
]

x = np.arange(len(quality_labels))
width = 0.35

plt.figure(figsize=(12, 6))

plt.bar(
    x - width/2,
    scorecard.iloc[0][quality_cols],
    width,
    label="Smithsonian"
)

plt.bar(
    x + width/2,
    scorecard.iloc[1][quality_cols],
    width,
    label="Met"
)

plt.xticks(x, quality_labels)
plt.ylabel("Coverage / Score (%)")
plt.title("Digital Heritage Data Quality Comparison")
plt.ylim(0, 100)
plt.legend()

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 3. GEOGRAPHIC DIVERSITY
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

plt.bar(
    scorecard["museum"],
    scorecard["unique_geographies"]
)

plt.title("Geographic Diversity")
plt.ylabel("Unique Standardized Geographies")
plt.xticks(rotation=15)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 4. GEOGRAPHIC CONCENTRATION
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

plt.bar(
    scorecard["museum"],
    scorecard["top_5_share"]
)

plt.title("Geographic Concentration — Top 5")
plt.ylabel("Share of Geographic Records (%)")
plt.ylim(0, 100)
plt.xticks(rotation=15)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 5. COLLECTION SCALE
# ------------------------------------------------------------

plt.figure(figsize=(10, 6))

plt.bar(
    scorecard["museum"],
    scorecard["objects"]
)

plt.title("Collection Scale")
plt.ylabel("Objects")
plt.xticks(rotation=15)

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 6. SCORECARD TABLE
# ------------------------------------------------------------

dashboard_table = scorecard[
    [
        "museum",
        "objects",
        "metadata_quality_score",
        "digital_access_score",
        "artist_coverage_score",
        "culture_coverage_score",
        "country_coverage_score",
        "unique_geographies",
        "top_5_share",
        "heritage_intelligence_score"
    ]
].copy()

dashboard_table.columns = [
    "Museum",
    "Objects",
    "Metadata Quality",
    "Digital Access",
    "Artist Coverage",
    "Culture Coverage",
    "Country Coverage",
    "Unique Geographies",
    "Top 5 Geographic Share",
    "Heritage Intelligence Score"
]

display(dashboard_table)


print("=" * 80)
print("FINAL DASHBOARD COMPLETE")
print("=" * 80)
```
